## 비선형 활성화 함수

- 직선 1개로는 그릴 수 없는 함수를 뜻함
- 선형 함수로는 은닉층을 여러 번 추가하더라도 1회 추가한 것과 차이를 줄 수 없음

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 시그모이드 함수 그래프를 그리는 코드
def sigmoid(x):
  return 1 / (1 + np.exp(-x))

x = np.arange(-5.0, 5.0, 0.1)
y = sigmoid(x)

plt.plot(x, y)
plt.plot([0, 0], [1.0, 0.0], ':')
plt.title('Sigmoid Function')
plt.show()
```

- 시그모이드 함수는 극단값에 갈수록 0에 가까운 아주 작은 값이 나옴
- 역전파 과정에서 0에 가까운 작은 값이 곱해지게 되면 앞단에는 기울기가 잘 전달되지 않게됨 -> 기울기 소실 문제

image.png

```python
from matplotlib.lines import lineStyles
# 하이퍼볼릭 탄젠트 함수
x = np.arange(-5.0, 5.0, 0.1)
y = np.tanh(x)

plt.plot(x, y)
plt.plot([0,0], [1.0, -1.0], ':')
plt.axhline(y=0, color='orange', linestyle='--')
plt.title('Tanh Function')
plt.show()
```

- -1과 1에 가까운 값을 출력할 때 시그모이드 함수와 같은 문제 발생함
- 시그모이드와 달리 0을 중심으로 하기 때문에 기울기 소실 증상이 상대적으로 적음
- 은닉층에서 시그모이드 함수보다는 많이 사용됨

```python
# ReLU  함수
def relu(x):
  return np.maximum(0, x)

x = np.arange(-5.0, 5.0, 0.1)
y = relu(x)

plt.plot(x, y)
plt.plot([0, 0], [5.0, 0.0], ':')
plt.title('ReLU Function')
plt.show()
```

- 시그모이드/ 하이퍼볼릭탄젠트와 달리 연산이 필요없고 단순 임계값이므로 연산 속도가 매우 빠름
- 입력값이 0이면 기울기도 0 -> 이 뉴런은 회생하는 것이 어려움 (dying ReLU)

```python
a = 0.1
```

```python
# 리키 렐루 (Leaky ReLU)
def leaky_relu(x):
  return np.maximum(a * x, x)

x = np.arange(-5.0, 5.0, 0.1)
y = leaky_relu(x)

plt.plot(x, y)
plt.plot([0, 0], [5.0, 0.0], ':')
plt.title('Reaky ReLU Function')
plt.show()
```

- 입력값이 음수라도 기울기가 0이 되지 않으면 렐루가 죽지 않음

```python
# 소프트맥스 함수
x = np.arange(-5.0, 5.0, 0.1)
y = np.exp(x) / np.sum(np.exp(x))

plt.plot(x, y)
plt.title('Softmax Function')
plt.show()
```

- 출력층에 적용하여 사용함 -> 분류 문제에 적용
