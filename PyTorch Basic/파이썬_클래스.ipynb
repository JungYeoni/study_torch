## 1. 함수(function)과 클래스(Class)의 차이

### 1-1. 함수로 덧셈기 구현하기

````python
# 전역 변수 선언
result = 0

def add(num):
  global result
  result += num
  return result
````

````python
print(add(3))
print(add(4))
````

### 1-2 함수로 두 개의 덧셈기 구현하기

````python
# 전역 변수 선언
result1 = 0
result2 = 0

def add1(num):
  global result1
  result1 += num
  return result1

def add2(num):
  global result2
  result2 += num
  return result2

print(add1(3))
print(add1(4))
print(add2(3))
print(add2(7))
````

### 1-3. 클래스(class)로 덧셈기 구현하기

````python
class Calculator:
  def __init__(self): # 객체 생성 시 호출될 때 수행되는 초기화 함수 -> 생성자 라고함
    self.result = 0

  def add(self, num): # 객체 생성 후 사용할 수 있는 함수
    self.result += num
    return self.result
````

````python
# 두 개의 객체 생성
cal1 = Calculator()
cal2 = Calculator()
````

````python
print(cal1.add(3))
print(cal1.add(4))
print(cal2.add(3))
print(cal2.add(7))
````

- 두 객체가 독립적으로 연산되고 있음을 확인할 수 있음

2-4. 곱하기, 빼기, 나누기 기능 만들기

````python
class FourCal:
  def setdata(self, first, second):
    self.first = first # 메서드의 수행문
    self.second = second # 메서드의 수행

  def add(self):
    result = self.first + self.second
    return result

  def mul(self):
    result = self.first * self.second
    return result

  def sub(self):
    result = self.first - self.second
    return result

  def div(self):
    result = self.first / self.second
    return result
````

2-5. 생성자

- 생성자: 객체가 생성될 때 자동으로 호출되는 메서드

````python
a = FourCal()
a.add()
````

- 객체에 초깃값을 설정해야 할 필요가 있을 때는 메서드를 호출하여 설정하기 보단 생성자를 구현하는 것이 안전함.

````python
class FourCal:
  def __init__(self, first, second):
    self.first = first
    self.second = second

  def add(self):
    result = self.first + self.second
    return result

  def mul(self):
    result = self.first * self.second
    return result

  def sub(self):
    result = self.first - self.second
    return result

  def div(self):
    result = self.first / self.second
    return result
````

````python
a = FourCal()
a.add()
````

````python
a = FourCal(4, 2)
a.add()
````

2-6. 클래스의 상속

-  어떤 클래스를 만들 때 다른 클래스의 기능을 물려받을 수 있도록 하는 것
- 기존 클래스를 변경하지 않고 기능을 추가하거나 기존 클래스를 변경하려고 할 때 사용한다.

*기존 클래스가 라이브러리 형태로 제공되거나 수정이 허용되지 않는 상황일 때 사용

````python
# 클래스를 상속하기 위해서는 클래스 이름 뒤 괄호 안에 상속할 클래스 이름을 넣어주면 됨
class MoreFourCal(FourCal):
  pass
````

````python
# MoreFourCal 클래스는 FourCal 클래스의 모든 기능을 사용할 수 있음
a = MoreFourCal(4, 2)

print(a.add())
print(a.mul())
print(a.sub())
print(a.div())
````

````python
class MoreFourCal(FourCal):
  def pow(self):
    result = self.first ** self.second
    return result
````

````python
a = MoreFourCal(4, 2)
print(a.pow())
print(a.add())
````

2-7. 메서드 오버라이딩

````python
a = MoreFourCal(4, 0)
a.div()
````

- 0으로 나눌 때 오류가 아닌 값을 리턴받고싶다면?

````python
class SafeFourCal(FourCal):
  def div(self):
    if self.second == 0: # 나누는 값이 0인 경우 0을 리턴
      return 0
    else:
      return self.first / self.second
````

````python
a = SafeFourCal(4, 0)
print(a.div())
````

2-8. 클래스 변수

- 클래스에 선언한 변수를 클래스 변수라고 함

````python
class Family:
  lastname = "김"
````

````python
Family.lastname
````

````python
a = Family()
b = Family()

print(a.lastname)
print(b.lastname)
````

````python
# 클래스 변수 바꾸기
Family.lastname = "박"

print(a.lastname)
print(b.lastname)
````

````python
# 클래스변수와 동일한 이름의 객체변수를 생성 -> a객체에 lastname 이라는 변수가 새롭게 생섣됨
a.lastname = "최"
a.lastname
````
